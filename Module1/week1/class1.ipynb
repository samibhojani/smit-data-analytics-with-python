```python
#Create a number guessing game.
import random

number = random.randint(1, 10)
tries = 1
while tries < 3:
    guessed_number = int(input('Enter number between 1 to 10: '))
    if guessed_number == number:
        print('Correct guess! you won!')
        tries = 4
        break
    elif guessed_number < number:
        print('Wrong guess! HINT: go a little higher.')
        tries += 1
    elif guessed_number > number:
        print('Wrong guess! HINT: go a little lower.')
        tries += 1
    else: print('Enter only a number between 1 to 10.')

if tries == 3 and guessed_number != number:
    print(f'You lose! correct numbe is {number}')
```

```python
#Create a simple login attempt system.
login_id = 'admin'
login_pswd = 'admin@12345'
attempts = 0
while attempts < 3:
    usr_login = input('Login ID: ')
    usr_pswd = input('Password: ')
    if usr_login == login_id and usr_pswd == login_pswd:
        print('Successful Login: Welcome to the Portal!')
        break
    elif usr_login == login_id and usr_pswd != login_pswd:
        print('ERROR: Invalid password!')
        attempts = attempts + 1
    elif usr_login != login_id and usr_pswd == login_pswd:
        print('ERROR: Invalid login!')
        attempts = attempts + 1
```

```python
#Create an ATM withdrawal loop.
acc_balance = 10000
while acc_balance != 0:
    withdrawal_amount = int(input('Enter Withdrawal amount'))
    acc_balance = acc_balance - withdrawal_amount
    print(f'Withdrawal Amount: {withdrawal_amount} PKR.\nRemaining Balance: {acc_balance}')
if acc_balance == 0:
    print('ACCOUNT EMPTY: All amount is withdrawed.')
```
